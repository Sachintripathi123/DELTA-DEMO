```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    "Name" : ["Nikhil", "sumit","Tom","prem","Sachin","Gaurav","Alen","pw"],
    "Income" : [20000,100000,22000,120000,150000,24000,125000,19000],
    "Spending" : [1000,5000,1100,8000,50000,18000,55000,16000]
}
df = pd.DataFrame(data)
df
```

```python
plt.scatter(df["Spending"],df["Income"])
plt.legend()
plt.grid(True)
plt.show()
```

```python
X = df[['Spending' , 'Income']]

```

```python
wcss = []

for k in range(1,9):
  kmeans = KMeans(n_clusters = k , init = "k-means++")
  kmeans.fit(X)
  wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,9), wcss)
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.title('Elbow Method For Optimal k')
plt.xticks(range(1,9))
plt.grid(True)
plt.show()
```

```python
plt.plot(range(1,9),wcss)
plt.xticks(range(1,9))
```

```python
from kneed import KneeLocator

knee = KneeLocator(range(1,9),wcss,curve = 'convex', direction="decreasing")
knee.elbow
```

```python
model = KMeans(n_clusters = 4)
df["group"] = model.fit_predict(X)

for i in df["group"].unique():
  group_data = df[df["group"] == i]
  plt.scatter(group_data["Spending"],group_data["Income"])
plt.legend()
plt.grid(True)
plt.show()
df
```

```python
new_income = int(input('enter your income:'))
new_spending = int(input('enter your spending:'))
prediction = model.predict([[new_spending,new_income]])

print(f"The new customer belongs to cluster: {prediction[0]}")
```
